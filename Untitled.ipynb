```python
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "pyproject.toml").exists():
    raise RuntimeError("Run this notebook from the HERALD repo root")

DATA = ROOT / "data"
RUN_ID = "010626_201904"  # set yours, or None to auto-pick latest

if RUN_ID is None:
    runs = sorted(
        (p for p in DATA.iterdir() if p.is_dir() and not p.name.startswith(".")),
        key=lambda p: p.stat().st_mtime,
        reverse=True,
    )
    if not runs:
        raise FileNotFoundError("No runs under data/ — run: uv run python scripts/scene_init.py …")
    RUN_ID = runs[0].name

print("Run id:", RUN_ID)
```

```python
from herald.data import RunPaths

paths = RunPaths(run_id=RUN_ID)

required = {
    "roi": paths.roi,
    "osm": paths.osm,
}
optional_vlm = {
    "aerial": paths.aerial,
    "aerial_meta": paths.aerial_meta,
}

for name, p in required.items():
    if not p.exists():
        raise FileNotFoundError(f"Missing {p}")
    print(name, "->", p)

for name, p in optional_vlm.items():
    print(name, "->", p, ("OK" if p.exists() else "missing (template-only mode)"))
```

```python
import json

from PIL import Image

from herald.scene.common.geometry import Frame
from herald.scene.common.roi import ROI
from herald.scene.init.io import raw_polygons_from_geojson
from herald.data.paths import FRAME_JSON
from services.aerial import AerialMeta

roi = ROI.from_geojson(paths.roi)
print(f"ROI area: {roi.area() / 1e6:.3f} km²")

with paths.osm.open(encoding="utf-8") as f:
    osm_geojson = json.load(f)
raw_polygons = raw_polygons_from_geojson(osm_geojson)
print(f"Raw OSM polygons: {len(raw_polygons)}")

frame = Frame.load(FRAME_JSON)
if frame is None:
    lat, lon = roi.latlon_centroid()
    frame = Frame.from_origin(lat, lon)
print(f"Frame origin: ({frame.lat:.5f}, {frame.lon:.5f})")

aerial_image = None
aerial_meta = None
if paths.aerial.exists() and paths.aerial_meta.exists():
    aerial_image = Image.open(paths.aerial).convert("RGB")
    meta_dict = json.loads(paths.aerial_meta.read_text(encoding="utf-8"))
    aerial_meta = AerialMeta(
        bbox=tuple(meta_dict["bbox"]),
        zoom=int(meta_dict["zoom"]),
        width_px=int(meta_dict["width_px"]),
        height_px=int(meta_dict["height_px"]),
        provider=str(meta_dict["provider"]),
    )
    print(f"Aerial: {aerial_image.size[0]}×{aerial_image.size[1]} px, zoom={aerial_meta.zoom}")
    print(f"bbox (S,W,N,E): {aerial_meta.bbox}")
else:
    print("No aerial mosaic — OSM template fallback only")
```

```python
from herald.scene.init.pipeline import seed_graph
from herald.scene.init.hierarchy import build_tree

graph = seed_graph(
    raw_polygons,
    roi,
    frame,
    emb_model_id="stub-v0",
    vlm_model_id="notebook-test",
)
build_tree(graph, frame)

polygons = [n for n in graph.nodes if n.type != "site"]
print(f"Graph nodes: {len(graph.nodes)} (site + {len(polygons)} polygons)")
print(f"Contains edges: {sum(1 for e in graph.edges if e.edge_type == 'contains')}")
```

```python
from services.aerial import fetch_osm_for_meta

osm_map_image = None
if aerial_meta is not None:
    print("Fetching OSM map aligned to aerial mosaic…")
    osm_map_image = fetch_osm_for_meta(aerial_meta, show_progress=True)
    if aerial_image is not None:
        assert osm_map_image.size == aerial_image.size, (osm_map_image.size, aerial_image.size)
    print(f"OSM map: {osm_map_image.size[0]}×{osm_map_image.size[1]} px")
else:
    print("Skipped OSM map (no aerial meta)")
```

```python
node = graph.get_node("node_30025278")

tags = {}
for ref in node.refs:
    if ref.assigned_by == "osm":
        tags = dict(ref.metadata.get("tags") or {})
        break

print(node.id, "| area:", node.refs[0].metadata.get("area") if node.refs else "?")
print("tags:", {k: tags[k] for k in sorted(tags)[:8]}, ("…" if len(tags) > 8 else ""))
```

```python
import importlib
import herald.scene.init.classify as classify_mod
importlib.reload(classify_mod)

from herald.scene.init.classify import (
    _classify,
    _crop_box,
    _crop_highlight,
    _ctx,
    _fallback,
    _name,
    _tags,
)
```

```python
tags = _tags(node)
ctx = {
    "area": round(float(node.refs[0].metadata.get("area", 0)), 1) if node.refs else 0,
    "tags": _ctx(tags),
    "name": _name(tags),
}

bbox, ring_local = _crop_box(
    node, frame, aerial_meta.bbox, aerial_meta.zoom,
    aerial_image.width, aerial_image.height,
)
L, T, R, B = bbox
ring_mosaic = [(x + L, y + T) for x, y in ring_local]

aerial_crop = _crop_highlight(aerial_image, bbox, ring_local)
osm_crop = _crop_highlight(osm_map_image, bbox, ring_local)

print("crop:", bbox)
print("ctx:", ctx)
```

```python
%matplotlib inline

import matplotlib.pyplot as plt
from PIL import Image, ImageDraw

_RGB = (255, 40, 40)

def highlight_on_mosaic(image, ring_px):
    out = image.convert("RGBA")
    layer = Image.new("RGBA", out.size, (0, 0, 0, 0))
    draw = ImageDraw.Draw(layer)
    draw.polygon(ring_px, fill=(*_RGB, 140), outline=(*_RGB, 255), width=3)
    return Image.alpha_composite(out, layer).convert("RGB")

aerial_full = highlight_on_mosaic(aerial_image, ring_mosaic)
osm_full = highlight_on_mosaic(osm_map_image, ring_mosaic)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes[0, 0].imshow(aerial_full); axes[0, 0].set_title("Full aerial"); axes[0, 0].axis("off")
axes[0, 1].imshow(osm_full);    axes[0, 1].set_title("Full OSM");    axes[0, 1].axis("off")
axes[1, 0].imshow(aerial_crop); axes[1, 0].set_title("Aerial crop (VLM)"); axes[1, 0].axis("off")
axes[1, 1].imshow(osm_crop);    axes[1, 1].set_title("OSM crop (VLM)");    axes[1, 1].axis("off")
plt.tight_layout()
plt.show()
```
